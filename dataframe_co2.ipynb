```python
counts_per_year = df_ch['Jahr'].value_counts()
max_points = counts_per_year.max()
years_with_max = counts_per_year[counts_per_year == max_points].index
start_jahr = years_with_max.min()
print(years_with_max)

```

```python
g = df_ch['Jahr'].value_counts().sort_index().plot(ylabel='Anzahl Datenpunkte', title='Anzahl Datenpunkte pro Jahr');
g.axvline(x=2013, color='violet',label='2013') 
g.legend() 
print(f"Ab dem Jahr {start_jahr} haben alle {max_points} Länder einen Datenpunkt.")
```

```python
g = df_ch['Emission pro Kopf'].plot(kind='hist', bins=20, title='Histogramm von Emission pro Kopf', 
                           xlabel='Emission pro Kopf', ylabel='Anzahl Datenpunkte')
mean = df_ch['Emission pro Kopf'].mean()
g.axvline(x=df_ch['Emission pro Kopf'].mean(), color='pink',label=f'Mittelwert = {mean:.2f}')  

median = df_ch['Emission pro Kopf'].median()
g.axvline(x=df_ch['Emission pro Kopf'].median(), color='green',label=  f'Median = {median:.2f}')  

g.legend()
```

```python
g = df_ch['BIP'].plot(kind='hist', bins=20, title='Histogramm von BIP', 
                           xlabel='BIP', ylabel='Anzahl Datenpunkte')
mean = df_ch['BIP'].mean()
g.axvline(x=df_ch['BIP'].mean(), color='yellow',label=f'Mittelwert = {mean:.2f}')  

median = df_ch['BIP'].median()
g.axvline(x=df_ch['BIP'].median(), color='red',label=  f'Median = {median:.2f}')  

g.legend()
```
