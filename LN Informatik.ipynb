## Projektidee
Wir untersuchen, ob reichere Länder pro Kopf mehr CO₂ ausstossen als ärmere. Dazu vergleichen wir das BIP pro Kopf mit dem CO₂-Ausstoss pro Kopf weltweit. Für auffällige Länder recherchieren wir mögliche Gründe.
